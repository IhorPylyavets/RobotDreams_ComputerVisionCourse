```python
import cv2
import matplotlib.pyplot as plt
```

```python
#open video
cap = cv2.VideoCapture("../data/slow_traffic.mp4")
#first frame
ret, frame = cap.read()

if ret:
    frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    plt.imshow(frame_rgb)
    plt.axis("off")
    plt.show()
else:
    print("Can't open")

cap.release()
```

```python
x1, y1 = 222, 121
x2, y2 = 278, 169
cv2.rectangle(frame_rgb, (x1, y1), (x2, y2), (0, 255, 0), 2)

x3, y3 = 259, 73
x4, y4 = 308, 121
cv2.rectangle(frame_rgb, (x3, y3), (x4, y4), (255, 0, 0), 2)

image_rgb = cv2.cvtColor(frame_rgb, cv2.COLOR_BGR2RGB)

plt.imshow(image_rgb)
plt.axis("off")
plt.show()
```
