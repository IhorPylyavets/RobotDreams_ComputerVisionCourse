Квантування зображення — це процес зменшення кількості можливих значень інтенсивності (або кольорів) у зображенні. 
Суть:
У цифровому зображенні кожен піксель може мати дуже багато значень (наприклад, у 8-бітному зображенні 256 рівнів яскравості, у 24-бітному — понад 16 млн кольорів). Квантування зменшує це число до меншого набору (наприклад, 16 відтінків сірого чи 256 кольорів).
- втрачаємо інформацію

Dithering — це метод, який додає "шумоподібні" точки при зменшенні кількості кольорів або відтінків, щоб уникнути різких переходів.
Ідея: замість того, щоб різко обрізати значення до найближчого кольору з палітри, ми розподіляємо похибку між сусідніми пікселями, створюючи ілюзію плавного переходу.
Як це працює? Якщо з 256 відтінків сірого зробити 4 рівні, то будуть лише "сходинки" і чіткі межі (постеризація).
Похибка між реальним і обраним значенням додається до сусідів. Це створює візуально більш плавні переходи.

У computer vision поняття sparsity (розрідженість): більшість даних можна представити у вигляді, де значна частина елементів дорівнює нулю або є неінформативною, а лише невелика кількість не-нульових (або "важливих") значень зберігає основну інформацію.

DCT (Discrete Cosine Transform — дискретне косинусне перетворення) - це лінійне перетворення, яке переводить сигнал (1D чи 2D, наприклад, зображення) з просторової області (пікселі) в частотну область (коефіцієнти косинусів різних частот).

Аналог Fourier Transform, але використовує лише косинусні функції (без синусів).

Важлива властивість: енергія сигналу після DCT концентрується в низькочастотних коефіцієнтах → більшість високочастотних можна занулити без великої втрати якості → це і є sparsity.

Приклад у Computer Vision

JPEG-компресія
-Зображення ділиться на блоки 8×8.
-До кожного блоку застосовується 2D DCT.
-Зберігаються тільки кілька найважливіших коефіцієнтів (низькі частоти), решта занулюються.
-Завдяки цьому зображення займає значно менше місця.
Фільтрація / денойзинг
-Шум в основному у високочастотних коефіцієнтах.
-Їх можна відсікати → чистіше зображення.
Розпізнавання образів
-DCT-ознаки використовуються для опису текстур, лицевих зображень (наприклад, Eigenfaces).
